```python
import jax.numpy as jnp
```

```python
A = jnp.arange(6400 * 6400).reshape((6400, 6400))/1000000 + 5 * jnp.eye(6400)
```

```python
import jax.random as random 
```

```python
key = random.key(0)
b = random.normal(key, shape=(6400,))
```

```python
b.block_until_ready
```

```python
%timeit jnp.linalg.solve(A, b).block_until_ready()
```
